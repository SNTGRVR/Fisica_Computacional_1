# <span style="color:red; font-style:italic;">Derivación</span>

Recordemos, desde la cinemática básica, si un cuerpo recorre una distancia de $\Delta x = 100\text{km}$ en un tiempo de $\Delta t = 2\text{ horas}$, su velocidad media es:

$$v_{\text{prom}} = \frac{\Delta x}{\Delta t} = \frac{100\text{ km}}{2\text{ h}} = 50\text{ km/h}\,.$$

Sin embargo, este valor no nos dice a qué velocidad avanzaba el vehículo en un instante exacto $t_0$. En realidad, el velocímetro marca distintos valores durante el trayecto.

Para medir el cambio instantáneo, reducimos el intervalo de tiempo $\Delta t$ haciéndolo cada vez más pequeño. La tasa de cambio en un instante exacto corresponde al límite:

$$v(t) = \lim_{\Delta t \to 0} \frac{x(t + \Delta t) - x(t)}{\Delta t} = \frac{dx}{dt}\,.$$

Geométricamente, la derivada es la pendiente de la recta tangente a la curva en un punto específico.

Suponga que se dispone de datos discretos (mediciones de tiempo $t$ y posición $x$). Usando la rutina `np.gradient`, cuyo manual puede [cosultar aquí](https://numpy.org/doc/stable/reference/generated/numpy.gradient.html) podemos obtener la tasa de cambio instantánea en cada punto, veamos

```python
# derivada_MRUA.py
import numpy as np

def x(t, x0=0., v0 = 0., a=2.5):
    'MRUA'
    return x0 + v0*t + 1/2. *a*t**2 

t = np.linspace(0,100,10)
x0 = float(input("Ingrese la posición inicial en m\n"))
v0 = float(input("Ingrese la velocidad inicial en m/s\n"))
a  = float(input("Ingrese la aceleración inicialen m0/s^2\n"))

r = x(t,x0,v0,a)

# velocidad v(t) = dx/dt
v_num = np.gradient(r, t)
# aceleración a(t) = dv/dt 
a_num = np.gradient(v_num, t)


v_num2 = np.gradient(r, t, edge_order=2)
a_num2 = np.gradient(v_num2, t, edge_order=2)

# velocidad analítica v(t) = v0 + a * t 
v_exacta = v0 + a*t
# aceleración analítica a(t) = a 
a_exacta = a    

print("\n" + "="*50)
print("Velocidad exacta:    \n", np.round(v_exacta, 2))
print("Velocidad numérica:  \n", np.round(v_num, 2))
print("-" * 50)
print("Aceleración exacta:  \n", a_exacta)
print("Aceleración numérica:\n", np.round(a_num, 2))
print("="*50)
print("Velocidad numérica edge_order2:  \n", np.round(v_num2, 2))
print("-" * 50)
print("Aceleración numérica edge_order2:\n", np.round(a_num2, 2))
print("="*50)
```

Cuando se trabaja con funciones en papel, la derivada se define mediante un límite infinitesimal. Sin embargo, numéricamente se trabaja con **muestras discretas** (mediciones tomadas a intervalos de tiempo $\Delta t$ o espacio $\Delta x$). La función `np.gradient()` de NumPy nos permite calcular la tasa de cambio instantánea (derivada numérica) a lo largo de un arreglo de datos utilizando el método de **diferencias finitas** (este lo estudiaremos en detalle en la continuación de este curso). El algoritmo calcula la derivada tomando el cuenta los vecinos hacia adelante y haci atrás. En los extremos, no se tienen vecinos y pr eso los valores numéricos no coinciden con los análiticos. Note que con el argumento `edge_order =2` corregimos el error.

En los puntos interiores de la malla, el algoritmo calcula diferencias centrales considerando vecinos hacia adelante y hacia atrás. En los extremos ($t_0$ y $t_{\text{final}}$) no existen ambos vecinos:
* Con `edge_order=1`, NumPy aproxima la frontera mediante diferencias hacia adelante/atrás de **primer orden** (ajuste lineal con 2 puntos).
* Con `edge_order=2`, utiliza una aproximación de **segundo orden** (ajuste parabólico con 3 puntos), reduciendo significativamente el error en los extremos.

### <span style="color:blue; font-style:italic;">Ejemplo: la derivada de $\sin(x)$</span>


```python
import numpy as np
import matplotlib.pyplot as plt

N =  50 # número de úntos
x = np.linspace(0,2*np.pi,N)

y = np.sin(x) 

dy = np.cos(x) # derivada exacta de sin(x)

# derivadas numéricas
dy1 =  np.gradient(y, x, edge_order=1) 
dy2 =  np.gradient(y, x, edge_order=2)

#errores
error1 = np.abs(dy - dy1)
error2 = np.abs(dy - dy2)

#plots
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 10), sharex=True)

# Subplot 1: Función y derivadas
ax1.plot(x, y, "k--", alpha=0.5, lw = 2,label=r"$f(x) = \sin(x)$")
ax1.plot(x, dy, "k-", lw=2, label=r"Analítica: $\cos(x)$")
ax1.plot(x, dy1, "ro", markersize=5, label=r"Numérica (edge_order=1)")
ax1.plot(x, dy2, "b.", markersize=4, label=r"Numérica (edge_order=2)")
ax1.grid(True, ls=":", alpha=0.6)
ax1.legend(loc="lower left", fontsize=18)
ax1.tick_params (axis='both', which='both', labelsize =20)

# Subplot 2: Error absoluto
ax2.semilogy(x, error1, "r-o", markersize=4, label=r"Error $|f'_{exacta} - f'_{num}|$ (edge=1)")
ax2.semilogy(x, error2, "b-s", markersize=4, label=r"Error $|f'_{exacta} - f'_{num}|$ (edge=2)")
ax2.set_xlabel(r"$x$ (rad)", fontsize=20)
ax2.set_ylabel("Error absoluto", fontsize=20)
ax2.grid(True, which="both", ls=":", alpha=0.6)
ax2.legend(loc="lower center", fontsize=18)
ax2.tick_params (axis='both', which='both', labelsize =20)

plt.tight_layout()
#plt.savefig('../lectures/03_herramientas_python/derivadasinx.pdf', bbox_inches='tight')
plt.show()
```

# <span style="color:red; font-style:italic;">Integración</span>

Consideremos ahora el problema inverso: conocemos la velocidad $v(t)$ en cada instante y queremos saber qué distancia total $x$ se recorrió. Si la velocidad fuese constante, bastaría una multiplicación simple $x = v \cdot \Delta t$, que geométricamente equivale al área de un rectángulo, (figura 2a). Pero si la velocidad cambia con el tiempo (figura 2b), la multiplicación directa deja de funcionar.

<p align="center">
  <img src="https://raw.githubusercontent.com/longas14/Fisica_Computacional_1/main/notebooks/figuras/integracion.png" width="420" alt="Velocidad constante: Área rectangular">
  <br>
  <em><b>Figura 1:</b> Área bajo una curva de velocidad constante.</em>
</p>


Para calcular el total acumulado si la velocidad es variable:
1. Dividimos el intervalo continuo en subintervalos muy pequeños de ancho $\Delta t$.
2. Suponemos que en cada subintervalo la velocidad $v(t_i)$ es aproximadamente constante.
3. Sumamos el área de los delgados rectángulos resultantes: $x \approx \sum v(t_i) \Delta t$.

Al hacer los intervalos infinitesimalmente delgados ($\Delta t \to 0$), la suma discreta se transforma en una integral definida:

$$
x_{\text{total}} = \int_{a}^{b} v(t) \, dt
$$

Geométricamente, la integral representa el área bajo la curva $v(t)$ entre los límites $t = a$ y $t = b$. Podríamos decir, sin entrar en sutilezas matemáticas, que el cálculo se sintetiza en la relación fundamental entre ambas operaciones: "la integral deshace la derivada y viceversa".

$$
\int_{a}^{b} \frac{dx(t)}{dt} \, dt = x(b) - x(a)
$$

### <span style="color:blue; font-style:italic;">Ejemplo: Integración de $f(x) = x$</span>
Esta integral es trivial y puede hacerse a "mano", note que si derivamos la funciópn $F(x) = \frac{1}{2}x^2 \implies F'(x) = x =f(x)$. Por el teorema fundamental del cálculo:

$$
\int_a^b f(x)\,dx = \int_a^b F'(x)\,dx = \left. F(x) \right|_{a}^{b} = \left. \frac{x^2}{2} \right|_{a}^{b} = \frac{b^2}{2} - \frac{a^2}{2}
$$

Para $x \in [0,2]$:

$$
I = \int_{0}^{2} x \, dx = \left. \frac{x^2}{2} \right|_{0}^{2} = \frac{4}{2} - 0 = 2
$$


```python
import scipy as sp
import numpy as np

f = lambda x: x

a = 0.; b=2.

integral, error = sp.integrate.quad(f, a, b)

print("Integral:", integral)
print("Error:   ", error)
```

### <span style="color:blue; font-style:italic;">Ejemplo: Integración de $f(x) = x^4 - 2x +1$</span>
En este caso, también se puede hacer a mano:

$$
\int_0^2 (x^4 - 2x +1) dx = \left. \frac{x^5}{5} - x^2 + x\right|_{0}^{2} = 4.4
$$

```python
import scipy as sp

f = lambda x: x**4 - 2*x + 1

a = 0
b = 2

integral, error = sp.integrate.quad(f, a, b)

print("Integral:", integral)
print("Error:   ", error)
```

### <span style="color:blue; font-style:italic;">Ejemplo: integral de $f(x) = e^{-x^2}$</span>
Consideremos la función $f(x) = e^{-x^2}$ en el intervalo $[0, 1]$. Esta integral no posee primitiva expresable en términos de funciones elementales y define la función de error:

$$
\int_0^1 e^{-x^2}\,dx = \frac{\sqrt{\pi}}{2} \operatorname{erf}(1) \approx 0.746824132812427
$$

En este tipo de problemas, el cálculo numérico no es solo una alternativa conveniente, sino una necesidad analítica.

```python
import scipy as sp
import numpy as np

f = lambda x: np.exp(-x**2)

integral, error = sp.integrate.quad(f, 0.0, 1.0)

print("Integral:", integral)
print("Error:   ", error)
```

### <span style="color:blue; font-style:italic;">Ejemplo: integral de $f(x) = e^{-x^2}$</span>

`scipy.integrate.quad` nos permite evaluar integrales impropias de primera especie, es decir, aquellas donde uno o ambos límites de integración se extienden al infinito:

$$\int_{a}^{\infty} f(x)\,dx, \qquad \int_{-\infty}^{b} f(x)\,dx, \qquad \int_{-\infty}^{\infty} f(x)\,dx$$

Bajo el capó, la subrutina `QAGI` de QUADPACK no discretiza un intervalo infinito arbitrario, sino que aplica transformaciones de variable racionales (por ejemplo, $x = \frac{1 - t}{t}$ para mapear $[0, \infty)$ en $[0, 1]$) y evalúa la función mediante cuadraturas adaptativas de Gauss-Kronrod, eliminando problemas de divergencia por muestreo. En Python, indicamos los límites infinitos mediante la constante `np.inf` o `-np.inf`.

Por ejemplo, consideremos la integral:

$$I = \int_{-\infty}^{\infty} e^{-x^2}\,dx = \sqrt{\pi} \approx 1.772453850905516$$

```python
import scipy as sp
import numpy as np

f = lambda x: np.exp(-x**2)

integral, error = sp.integrate.quad(f, -np.inf, np.inf)

print("Integral:", integral)
print("Error:   ", error)
```

### <span style="color:blue; font-style:italic;">Ejemplo: integral de datos</span>
Suponga que en un archivo `velocidades.txt` tiene guardado datos experimentales. Estos representan el tiempo y la velocidad. Recordando que la velocidad y la aceleración están dadas por $v = \dfrac{dx}{dt} \implies x = \int v dt$, $a = \dfrac{dv}{dt}$, podemos integrar y derivar los datos para obtener estas variables físicas.

```python
# integral_datos.py
import numpy as np
import scipy as sp
import matplotlib.pyplot as plt

# Cargar datos del archivo
datos = np.loadtxt("../scripts/velocidades.txt")
t = datos[:, 0]  # Tiempo en s
v = datos[:, 1]  # Velocidad en m/s

# Integración: trayectoria acumulada 
# x(t) = integral_0^t v(t') dt'
# initial=0.0 asegura x(t0) = 0 y mantiene la misma dimensión que t
x_acum = sp.integrate.cumulative_trapezoid(v, x=t, initial=0.0)

# Distancia total recorrida
dist_total = sp.integrate.simpson(v, x=t)
print(f"Distancia total recorrida (método de Simpson): {dist_total:.4f} m")

# Derivación: aceleración instantánea a(t) = dv/dt
a_num = np.gradient(v, t, edge_order=2)

# Plots
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(8, 9), sharex=True)

# Panel 1: Posición acumulada
ax1.plot(t, x_acum, "tab:red", lw=2, marker="s", markersize=4, label="Posición $x(t) = \int v dt$")
ax1.set_ylabel("x (m)", fontsize=11)
ax1.grid(True, ls=":", alpha=0.6)
ax1.legend(loc="upper left")

# Panel 2: velocidad
ax2.plot(t, v, "tab:blue", lw=2, marker="o", markersize=4, label="Velocidad experimental $v(t)$")
ax2.set_ylabel("v (m/s)", fontsize=11)
ax2.grid(True, ls=":", alpha=0.6)
ax2.legend(loc="lower left")

# Panel 3: Aceleración 
ax3.plot(t, a_num, "tab:green", lw=2, marker="^", markersize=4, label=r"Aceleración $a(t) = \frac{dv}{dt}$ ")
ax3.set_xlabel("Tiempo (s)", fontsize=11)
ax3.set_ylabel("a (m/s$^2$)", fontsize=11)
ax3.grid(True, ls=":", alpha=0.6)
ax3.legend(loc="upper center")

#plt.savefig('../lectures/03_herramientas_python/integral_datos.pdf', bbox_inches='tight')

plt.tight_layout()
plt.show()
```

  <h4 style="color: #00695c; margin-top: 12px; margin-bottom: 4px;">1. <code>scipy.integrate.quad</code> (Cuadratura Adaptativa de Gauss-Kronrod)</h4>
  <p>Se utiliza cuando se conoce la <b>expresión matemática analítica</b> $f(x)$ y se desea evaluar $\int_a^b f(x)\,dx$.</p>
  <ul>
    <li><b><code>func</code> (Obligatorio):</b> Función ejecutable en Python que acepta como primer argumento la variable independiente $x$ y retorna un valor escalar $f(x)$.</li>
    <li><b><code>a, b</code> (Obligatorio):</b> Límites inferior y superior de integración. Pueden ser números reales o infinitos (<code>-np.inf</code>, <code>np.inf</code>).</li>
    <li><b><code>args</code> (Opcional, tupla):</b> Parámetros adicionales que requiera la función $f(x, p_1, p_2)$ fuera de la variable de integración. Ejemplo: <code>args=(k, m)</code>.</li>
    <li><b>Lo que devuelve:</b> Una tupla de dos valores: <code>(resultado, error_estimado)</code>, donde el segundo valor es una cota superior estricta del error absoluto numérico.</li>
  </ul>

  <h4 style="color: #00695c; margin-top: 12px; margin-bottom: 4px;">2. <code>scipy.integrate.trapezoid</code> (Regla del trapecio)</h4>
  <p>Aproxima el área bajo la curva uniendo puntos adyacentes con <b>segmentos lineales</b>. Ideal para datos experimentales o de sensores con variaciones suaves o pocos puntos.</p>
  <ul>
    <li><b><code>y</code> (Obligatorio):</b> Arreglo con las muestras dependientes ($f_i$).</li>
    <li><b><code>x</code> (Opcional):</b> Arreglo con la variable independiente. Es <b>imprescindible</b> si el muestreo no es uniforme (por ejemplo, tiempos variables $t_i$).</li>
    <li><b><code>dx</code> (Opcional, por defecto <code>1.0</code>):</b> Paso de integración escalar constante si no se especifica el vector <code>x</code>.</li>
    <li><b>Lo que devuelve:</b> Un número escalar con la estimación acumulada del área total.</li>
  </ul>

  <h4 style="color: #00695c; margin-top: 12px; margin-bottom: 4px;">3. <code>scipy.integrate.simpson</code> (Regla de Simpson)</h4>
  <p>Aproxima el área ajustando <b>parábolas</b> sobre tríos de puntos consecutivos. Ofrece mayor precisión que el trapecio para funciones continuas con curvatura significativa.</p>
  <ul>
    <li><b><code>y, x, dx</code>:</b> Mismo uso y firma de parámetros que en <code>trapezoid</code>.</li>
    <li><b>Requisito práctico:</b> Funciona mejor cuando los datos provienen de un muestreo uniforme (<code>dx</code> constante) y se dispone de un número impar de puntos (subintervalos pares).</li>
    <li><b>Lo que devuelve:</b> Un número escalar con la estimación numérica acumulada del área total.</li>
  </ul>

  <h4 style="color: #00695c; margin-top: 12px; margin-bottom: 4px;">4. <code>scipy.integrate.cumulative_trapezoid</code> (Integral Acumulada / Trayectoria)</h4>
  <p>Evalúa la integral acumulativa paso a paso como una función del límite superior: $F(t) = \int_{t_0}^t y(\tau)\,d\tau$. A diferencia de <code>trapezoid</code> (que devuelve un único valor escalar del área total), esta rutina entrega un arreglo con la evolución temporal o espacial de la integral.</p>
  <ul>
    <li><b><code>y</code> (Obligatorio, primer argumento):</b> Arreglo con la <b>variable dependiente</b> a integrar (datos experimentales, velocidades $v(t)$, aceleraciones $a(t)$, tasas de producción, etc.).</li>
    <li><b><code>x</code> (Opcional, segundo argumento):</b> Arreglo con la <b>variable independiente</b> (eje temporal $t$ o posición $r$). Es indispensable si el muestreo experimental no es regular ($\Delta t$ variable).</li>
    <li><b><code>dx</code> (Opcional, por defecto <code>1.0</code>):</b> Espaciado constante entre muestras si no se provee el arreglo <code>x</code>.</li>
    <li><b><code>initial</code> (Crítico para dimensiones):</b> Dado que $N$ mediciones definen únicamente $N-1$ intervalos de integración:
      <ul>
        <li><b><code>initial=None</code> (por defecto):</b> Devuelve un arreglo de longitud $N - 1$, omitiendo el instante inicial $t_0$. Si se intenta graficar directamente contra <code>x</code>, Python arrojará un error de dimensiones (<code>ValueError</code>).</li>
        <li><b><code>initial=0.0</code>:</b> Fija la condición $F(t_0) = \int_{t_0}^{t_0} y(\tau)\,d\tau = 0.0$. La salida conserva exactamente la longitud $N$, permitiendo graficar de inmediato pares $(t_i, F(t_i))$.</li>
      </ul>
    </li>
    <li><b>Lo que devuelve:</b> Un arreglo de NumPy con la serie integrada (por ejemplo, la trayectoria de posición $x(t)$ a partir de mediciones discretas de $v(t)$).</li>
  </ul>

  <h4 style="color: #00695c; margin-bottom: 4px;">Criterio de selección</h4>
  <ul>
    <li><b>¿Tienes la fórmula analítica $f(x)$?</b> $\longrightarrow$ Usa <code>scipy.integrate.quad</code>.</li>
    <li><b>¿Tienes datos discretos y buscas un valor total (escalar)?</b> $\longrightarrow$ Usa <code>trapezoid</code> o <code>simpson</code>.</li>
    <li><b>¿Tienes datos discretos y quieres graficar la curva integrada en el tiempo?</b> $\longrightarrow$ Usa <code>cumulative_trapezoid(..., initial=0)</code>.</li>
  </ul>
