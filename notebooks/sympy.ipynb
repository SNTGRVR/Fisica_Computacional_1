# <span style="color:#b22222; font-style:italic;">SymPy: Cómputo Simbólico </span>

Hasta el momento nos hemos enfocado exclusivamente en **métodos numéricos**. Sin embargo, en física y matemáticas aplicadas resulta fundamental abordar los problemas de manera **analítica** (o simbólica). Trabajar con soluciones cerradas permite:
* Hallar relaciones funcionales exactas sin error de truncamiento ni dispersión numérica.
* Estudiar regímenes asintóticos, límites físicos y expansiones perturbativas (series de Taylor).
* Obtener expresiones algebraicas explícitas para cantidades conservadas, ecuaciones de movimiento y matrices jacobianas.

Tradicionalmente, este tipo de tareas se delega a sistemas de álgebra computacional (**CAS**, por sus siglas en inglés) comerciales como *Wolfram Mathematica* o *Maple*, o a entornos abiertos como *SageMath*. En el ecosistema de Python, la librería por excelencia para el cálculo analítico exacto es **SymPy**.

A diferencia de las librerías numéricas convencionales:
1. **NumPy / SciPy:** Operan con arreglos de números en punto flotante (`float64`) con precisión finita.
2. **SymPy:** Modela variables, parámetros y operadores como objetos matemáticos formales abstractos, preservando fracciones exactas, raíces y constantes trascendentes ($\pi$, $e$) sin aproximar a decimales a menos que se solicite explícitamente.

Además, SymPy no busca reemplazar a NumPy, sino complementarlo: a través de herramientas como `sp.lambdify`, permite derivar analíticamente una solución exacta o ecuación diferencial y luego compilarla a código vectorial ultrarrápido para su evaluación y visualización numérica.

```python
import sympy as sy

sy.init_printing(use_latex='mathjax') # para usar latex en las variables
x = sy.symbols('x')
a,b,c = sy.symbols("a,b,c", real=True)
i,j,n = sy.symbols("i,j,n", integer=True)
theta = sy.symbols('theta')
```

```python
sy.cos(theta)**2 +x
```

# Álgebra básica: `expand`, `simplify`, `factor`, `collect`

```python
ecuacion = 5*x - 15
solucion= sy.solve(ecuacion, x)
print(f"la solución de la {ecuacion} = 0 es {solucion[0]}")
```

```python
cuadratica = a*x**2 + b*x + c 
sy.solve(cuadratica,x)
```

```python
factor = (x-a)*(x-b)
expan = sy.expand(factor)
print(expan)
```

```python
sy.collect(expan, a)
```

```python
factor.subs(x,c)
```

```python
sy.simplify(_)
```

# Cálculo: `limit`, `series`, `diff`, `integrate`, `Derivative`, `Integral`

```python
dfx = sy.diff(3*x**2 + 1,x) # derivada
ifx = sy.integrate(3*x**2 + 1,x) # integral
limite = sy.limit(sy.sin(x)/x,x,0)
print("si f(x) = 3x^2 + 1, su derivada es",dfx)
print("si f(x) = 3x^2 + 1, su integral es",ifx)
print("limite cuando x->0 de sinx/x=",limite)
```

```python
sy.integrate((x**2-1)/(x**3-1))
```

```python
sy.integrate((x**2-1)/(x**3-1), (x, 0, 1))
```

```python
sy.series(sy.sin(x), x)
```

# Álgebra lineal: `Matrix`, `det`, `eigenvals`, `eigenvects`

```python
omega = sy.symbols('omega', positive=True)
k = sy.symbols('k')
m = sy.symbols('m')
K = sy.Matrix([[2*k, -k], [-k, 2*k]])
M_mat = sy.Matrix([[m, 0], [0, m]])

# Ecuación secular det(K - omega^2 * M) = 0
ecuacion_secular = (K - omega**2 * M_mat).det()
modos = sy.solve(ecuacion_secular, omega**2)

print("la matrix:", K)
print(modos)
print(ecuacion_secular)
```

```python
K.eigenvals()
```

```python
K.eigenvects()
```

# Ecuaciones diferenciales

```python
y = sy.Function('y')
edo = sy.Eq(sy.diff(y(x), x, 2) + 4*y(x), 0)
sy.dsolve(edo, y(x))
```

### Una propiedad importante de sympy es que podemos convertir expresiones a funciones lambda y trabajar numéricamente con ellas.

```python
f_x = 5*x**3 + 2*x - 1
df_dx = sy.diff(f_x,x)

f = sy.lambdify([x], f_x)
dfdx = sy.lambdify([x], df_dx)

print(f(0), dfdx(0))
```
